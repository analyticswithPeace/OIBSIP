### Missing Data Handling Decisions

- **last_review:** Missing values were retained as NaN when `number_of_reviews` is 0. These listings have never received a review, so there is no valid last-review date to impute. Mean, median, mode, and forward-fill methods would create misleading dates.

- **reviews_per_month:** Missing values were replaced with 0 when `number_of_reviews` is 0. A value of 0 appropriately represents no review activity for listings that have never received a review.

- **name:** The missing values will be investigated before deciding whether to remove the affected rows, because the listing name is an important descriptive field.

- **host_name:** The missing values will be investigated before deciding on the appropriate treatment. Since this is a categorical field, numerical imputation methods such as mean or median are inappropriate.

### Standardization

Text-based columns were inspected for inconsistent formatting. Leading and trailing whitespace was removed from text fields using the `str.strip()` method. No arbitrary changes were made to legitimate names or location names, since differences in capitalization, punctuation, or wording may represent valid values.

```python
numeric_columns = df.select_dtypes(include=np.number).columns

numeric_columns
```

```python
outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()
    
    outlier_summary.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outliers': outliers
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

### Outlier Detection and Treatment

The Interquartile Range (IQR) method was used to identify potential outliers in the numerical columns.

The IQR method identified statistical outliers in several variables, including `host_id`, `latitude`, `longitude`, `price`, `minimum_nights`, `number_of_reviews`, `reviews_per_month`, and `calculated_host_listings_count`.

However, an IQR outlier is not necessarily an incorrect value. Many extreme values in this dataset can represent legitimate Airbnb listings, such as properties with high prices, longer minimum stays, many reviews, or hosts managing multiple listings.

Therefore:

- `id` and `host_id`: Retained because they are identifiers, and their magnitude does not represent an unusual measurement.
- `latitude` and `longitude`: Retained because extreme geographic values may represent legitimate locations.
- `price`: Extreme values will be retained unless they are confirmed to be invalid, because high-priced listings can be legitimate.
- `minimum_nights`: Extreme values will be retained because some legitimate listings may require long minimum stays.
- `number_of_reviews`: Extreme values will be retained because highly reviewed listings are legitimate.
- `reviews_per_month`: Extreme values will be retained because highly active listings can legitimately receive many reviews.
- `calculated_host_listings_count`: Extreme values will be retained because professional hosts may legitimately manage many properties.
- `availability_365`: No IQR outliers were detected.

No statistical outliers were removed or capped solely because they exceeded the IQR boundaries.

```python
print("Price <= 0:", (df['price'] <= 0).sum())
print("Minimum nights <= 0:", (df['minimum_nights'] <= 0).sum())
print("Availability outside 0-365:", ((df['availability_365'] < 0) | (df['availability_365'] > 365)).sum())
```

### Data Type Correction

The data types of all columns were reviewed to ensure they are appropriate for analysis.

- `id` was converted from integer to string because it is an identifier rather than a numerical measurement.
- `host_id` was converted from integer to string for the same reason.
- `price` was converted from integer to float because it represents a monetary value.
- `last_review` was converted from string to datetime because it contains dates.
- `latitude` and `longitude` were retained as float values because they represent geographic coordinates.
- `minimum_nights`, `number_of_reviews`, `calculated_host_listings_count`, and `availability_365` were retained as integers because they represent counts.
- `reviews_per_month` was retained as a float because it represents a numerical rate.
- Text columns were retained as string types.

```python
df['id'] = df['id'].astype(str)
df['host_id'] = df['host_id'].astype(str)
df['price'] = df['price'].astype(float)
df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')
```

### Missing Data Treatment for `name` and `host_name`

The `name` column contains 16 missing values and the `host_name` column contains 21 missing values. These fields are categorical/text variables, so mean and median imputation are not appropriate.

The missing values cannot be reliably reconstructed from the available data. Because only 37 rows are affected out of 48,895 records (approximately 0.08%), the affected rows were removed. This minimizes information loss while ensuring that the final dataset does not contain missing values in these important descriptive fields.

The `last_review` missing values were retained because they correspond to listings with zero reviews and therefore have no valid review date.
